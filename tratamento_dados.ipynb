```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'Dataset_Repatriados.CSV',encoding='latin-1',sep=';')
```

```python
df.info()
```

```python
#Definindo um função só pra traver as valores dos dados da variável
def contar(serie):
        return serie.value_counts(dropna=False)
```

Vamos criar um cópia do dataset para poder trabalhar com ela

```python
contar(df["Hipertensão"])
```

```python
#Criando um dataset de copia para poder trabalhar com ele

df_copia = df.copy()
```

```python
#Conferindo se o ID Pessoa é uma chave única para cada pessoa
df_copia["ID Pessoa"].value_counts().loc[lambda x : x >= 2]
```

```python
df_copia["Consentiu"] = df["Consentiu"].map({'Sim':1,'Não':0})
```

```python
df_copia["Boletim de Ocorrência/Sem Documento"].value_counts()
```

```python
df_copia["Boletim de Ocorrência/Sem Documento"]=df_copia["Boletim de Ocorrência/Sem Documento"].map({'Sim':1,'Não':0})
```

```python
df_copia["Boletim de Ocorrência/Sem Documento"].value_counts()
```

```python
df_copia["Você solicitou autodeportação/ deportação voluntária?"].value_counts()
```

```python
df_copia["Você solicitou autodeportação/ deportação voluntária?"] = df_copia["Você solicitou autodeportação/ deportação voluntária?"].map({'Sim':1,'Não':0})
```

```python
df_copia["Sem Comorbidade"].value_counts()
```

```python
#Tirando os espaços dos valores

df_copia["Sem Comorbidade"] = df_copia["Sem Comorbidade"].str.strip()
```

```python
#Usando o .replace para alterar valores especificos da variável é diferente de usar o metódo .map
df_copia["Sem Comorbidade"] = df_copia["Sem Comorbidade"].replace({'Sem Informação':'Sem informação',
                                                                    'Sem Comorbidade':'Sem comorbidade',
                                                                    'Não possui':'Sem comorbidade'})
```
